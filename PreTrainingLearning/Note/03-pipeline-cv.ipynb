# Hugging Face Transformers Pipeline - Computer Vision

Kaggle Notebook에서 바로 실행 가능한 `transformers.pipeline` 기반 이미지 분류 예제입니다.

사용 흐름:
1. `transformers` 라이브러리 설치
2. 이미지 URL에서 입력 이미지 로드
3. `pipeline("image-classification")` 생성
4. 모델로 이미지 분류 수행
5. 결과 출력

```python
# Kaggle Notebook에서 필요한 라이브러리 설치
!pip install -q transformers pillow requests
```

```python
from io import BytesIO

import requests
from IPython.display import display
from PIL import Image
from transformers import pipeline
```

## 1. 이미지 URL에서 입력 데이터 로드

```python
image_url = "https://images.unsplash.com/photo-1514888286974-6c03e2ca1dba?w=640"

response = requests.get(image_url, timeout=30)
response.raise_for_status()

image = Image.open(BytesIO(response.content)).convert("RGB")

display(image)
```

## 2. `pipeline("image-classification")` 생성

```python
# 기본 이미지 분류 모델을 명시해서 Kaggle에서도 재현 가능하게 실행합니다.
classifier = pipeline(
    "image-classification",
    model="google/vit-base-patch16-224",
)
```

## 3. 모델로 이미지 분류 수행 및 결과 출력

```python
results = classifier(image)

print("Image URL:", image_url)
print("\nTop predictions:")
for idx, item in enumerate(results, start=1):
    label = item["label"]
    score = item["score"]
    print(f"{idx}. {label} - {score:.4f}")
```

## 4. 여러 이미지 URL을 한 번에 분류하기

```python
image_urls = [
    "https://images.unsplash.com/photo-1514888286974-6c03e2ca1dba?w=640",
    "https://images.unsplash.com/photo-1552053831-71594a27632d?w=640",
    "https://images.unsplash.com/photo-1500530855697-b586d89ba3ee?w=640",
]

images = []
for url in image_urls:
    response = requests.get(url, timeout=30)
    response.raise_for_status()
    images.append(Image.open(BytesIO(response.content)).convert("RGB"))

for url, img in zip(image_urls, images):
    print(url)
    display(img)
    predictions = classifier(img, top_k=3)
    for idx, item in enumerate(predictions, start=1):
        print(f"{idx}. {item['label']} - {item['score']:.4f}")
    print("-" * 80)
```

## 5. 다른 이미지 URL로 테스트하기

`custom_image_url` 값만 바꾸면 Kaggle Notebook에서 다른 이미지를 바로 분류할 수 있습니다.

```python
custom_image_url = "https://images.unsplash.com/photo-1543466835-00a7907e9de1?w=640"

response = requests.get(custom_image_url, timeout=30)
response.raise_for_status()
custom_image = Image.open(BytesIO(response.content)).convert("RGB")

display(custom_image)

custom_results = classifier(custom_image, top_k=5)

print("Image URL:", custom_image_url)
print("\nTop predictions:")
for idx, item in enumerate(custom_results, start=1):
    print(f"{idx}. {item['label']} - {item['score']:.4f}")
```
