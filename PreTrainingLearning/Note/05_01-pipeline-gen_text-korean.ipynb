# Hugging Face Transformers Pipeline - Korean Prompt Text Generation

Kaggle Notebook에서 바로 실행 가능한 Hugging Face `transformers.pipeline` 기반 텍스트 생성 예제입니다.

이 노트북은 `gpt2` 모델을 사용하되, 입력 프롬프트는 영어 대신 한글 문장으로 넣어 텍스트를 생성합니다.

사용 조건:
1. `pipeline("text-generation")` 사용
2. `gpt2` 모델 사용
3. 한글 입력 프롬프트 기반 텍스트 생성
4. `max_length`, `num_return_sequences=8` 설정
5. 생성 결과 출력

참고: 기본 `gpt2`는 영어 중심으로 학습된 모델이라 한글 생성 품질은 제한적일 수 있습니다. 여기서는 요청 조건에 맞춰 모델은 `gpt2`로 고정합니다.

```python
# Kaggle Notebook에서 필요한 라이브러리 설치
# transformers는 Hugging Face 모델과 pipeline API를 사용하기 위해 필요합니다.
!pip install -q -U transformers
```

```python
# pipeline 함수를 불러옵니다.
import transformers
from transformers import pipeline

print("transformers version:", transformers.__version__)
```

## 1. 텍스트 생성 Pipeline 생성

`pipeline("text-generation")`와 `gpt2` 모델을 사용해 텍스트 생성기를 만듭니다.

```python
# text-generation pipeline을 생성합니다.
# model="gpt2"를 명시해서 GPT-2 기본 모델을 사용합니다.
text_generator = pipeline(
    "text-generation",
    model="gpt2",
)

# GPT-2에는 기본 pad_token이 없으므로 경고를 줄이기 위해 eos_token을 pad_token으로 사용합니다.
text_generator.tokenizer.pad_token = text_generator.tokenizer.eos_token
```

## 2. 한글 입력 프롬프트 준비

모델이 이어서 작성할 시작 문장을 한글로 입력합니다.

```python
# 텍스트 생성을 시작할 한글 입력 프롬프트입니다.
prompt = "인공지능은 미래의 교육을 다음과 같이 변화시킬 것이다."

print("한글 프롬프트:")
print(prompt)
```

## 3. GPT-2로 텍스트 생성

`max_length`로 전체 토큰 길이를 제한하고, `num_return_sequences=8`로 생성 결과 8개를 반환합니다.

```python
# 한글 프롬프트를 기반으로 텍스트를 생성합니다.
# max_length는 프롬프트를 포함한 최대 토큰 길이입니다.
# num_return_sequences=8은 서로 다른 생성 결과 8개를 반환한다는 의미입니다.
# do_sample=True를 사용해야 여러 개의 다양한 결과를 생성할 수 있습니다.
generated_outputs = text_generator(
    prompt,
    max_length=80,
    num_return_sequences=8,
    do_sample=True,
    temperature=0.9,
    top_k=50,
    top_p=0.95,
    pad_token_id=text_generator.tokenizer.eos_token_id,
)

print(f"생성된 문장 개수: {len(generated_outputs)}")
```

## 4. 생성 결과 출력

```python
# 생성된 8개 결과를 순서대로 출력합니다.
for idx, output in enumerate(generated_outputs, start=1):
    print(f"[생성 결과 #{idx}]")
    print(output["generated_text"])
    print("-" * 100)
```

## 5. 다른 한글 프롬프트로 테스트하기

`custom_prompt` 값만 바꾸면 다른 한글 주제로 텍스트를 생성할 수 있습니다.

```python
# 원하는 한글 프롬프트로 변경해서 실행해 보세요.
custom_prompt = "로봇과 사람이 함께 일하는 세상에서는"

custom_outputs = text_generator(
    custom_prompt,
    max_length=70,
    num_return_sequences=8,
    do_sample=True,
    temperature=0.8,
    top_k=50,
    top_p=0.95,
    pad_token_id=text_generator.tokenizer.eos_token_id,
)

print("사용자 한글 프롬프트:")
print(custom_prompt)
print("\n생성 결과:")

for idx, output in enumerate(custom_outputs, start=1):
    print(f"[사용자 프롬프트 생성 결과 #{idx}]")
    print(output["generated_text"])
    print("-" * 100)
```
